```python
from glob import glob
import pandas as pd
import numpy as np
import cv2
```

```python
data_dr = r"C:/Users/athet/Downloads/archive/kaggle_3m"
```

```python
def create_df(data_dir):
    images_paths = []
    masks_paths = glob(f'{data_dir}/*/*_mask*')

    for i in masks_paths:
        images_paths.append(i.replace('_mask', ''))

    df = pd.DataFrame(data= {'images_paths': images_paths, 'masks_paths': masks_paths})

    return df

```

```python
df = create_df(data_dr)
df.head()

```

```python
def df_to_arrays_raw(df):
    images = []
    masks = []

    for i in range(len(df)):
        img_path  = df.iloc[i]['images_paths']
        mask_path = df.iloc[i]['masks_paths']

        img  = cv2.imread(img_path)       # BGR, original resolution
        mask = cv2.imread(mask_path)      # BGR, original resolution

        images.append(img)
        masks.append(mask)

    return np.array(images), np.array(masks)
```

```python
X, Y = df_to_arrays_raw(df)

print("Total Images:", len(X))
print("Total Masks:", len(Y))

print("Images shape:", X.shape)
print("Masks shape:", Y.shape)

np.save('imagesegmentationimages.npy', X)
np.save('imagesegmentationmasks.npy', Y)
print("Saved imagesegmentationimages.npy and imagesegmentationmasks.npy locally.")
```
